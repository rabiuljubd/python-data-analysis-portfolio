```python
def cronbach_alpha(df):
    items = df.dropna()
    item_variance = items.var(axis=0, ddof=1)
    total_variance = items.sum(axis=1).var(ddof=1)
    n_items = items.shape[1]
    
    alpha = (n_items/(n_items-1)) * (1 - item_variance.sum()/total_variance)
    return alpha


for construct, items in construct_items.items():
    alpha = cronbach_alpha(df_pls[items])
    print(construct, "Cronbach Alpha =", round(alpha,3))
```

```python
import numpy as np

def calculate_cr_ave(df):
    corr = df.corr()
    loadings = np.sqrt(corr.mean(axis=1))
    
    CR = (loadings.sum()**2) / ((loadings.sum()**2) + ((1-loadings**2).sum()))
    AVE = (loadings**2).mean()
    
    return CR, AVE


for construct, items in construct_items.items():
    cr, ave = calculate_cr_ave(df_pls[items])
    print(construct)
    print("CR =", round(cr,3))
    print("AVE =", round(ave,3))
    print()
```

```python
import pandas as pd
import numpy as np

def calculate_vif(df):
    vif_data = pd.DataFrame()
    vif_data["Variable"] = df.columns
    
    vif_values = []

    for i in range(df.shape[1]):
        y = df.iloc[:, i]
        X = df.drop(df.columns[i], axis=1)

        X = np.column_stack([np.ones(len(X)), X])

        r_squared = np.linalg.lstsq(X, y, rcond=None)[1]

        # calculate R²
        y_pred = X @ np.linalg.lstsq(X, y, rcond=None)[0]
        ss_res = ((y - y_pred)**2).sum()
        ss_tot = ((y - y.mean())**2).sum()
        R2 = 1 - ss_res/ss_tot

        vif = 1/(1-R2)
        vif_values.append(vif)

    vif_data["VIF"] = vif_values
    
    return vif_data


calculate_vif(structural_df)
```

```python
import numpy as np

def calculate_r2(X, y):
    X_matrix = np.column_stack([np.ones(len(X)), X])
    
    beta = np.linalg.lstsq(X_matrix, y, rcond=None)[0]
    
    y_pred = X_matrix @ beta
    
    ss_res = ((y - y_pred) ** 2).sum()
    ss_tot = ((y - y.mean()) ** 2).sum()
    
    r2 = 1 - (ss_res / ss_tot)
    
    return r2


# FMB: FL + DFB
r2_FMB = calculate_r2(
    structural_df[["FL", "DFB"]],
    structural_df["FMB"]
)

# ISB: DFB + SI
r2_ISB = calculate_r2(
    structural_df[["DFB", "SI"]],
    structural_df["ISB"]
)

# FFO: FL + FMB + ISB
r2_FFO = calculate_r2(
    structural_df[["FL", "FMB", "ISB"]],
    structural_df["FFO"]
)


print("R² FMB =", round(r2_FMB, 3))
print("R² ISB =", round(r2_ISB, 3))
print("R² FFO =", round(r2_FFO, 3))
```

```python
import numpy as np
import pandas as pd


def bootstrap_path(X, y, n_boot=5000):

    beta_results = []

    n = len(X)

    for i in range(n_boot):

        # resample
        idx = np.random.randint(0, n, n)

        X_sample = X.iloc[idx]
        y_sample = y.iloc[idx]

        X_matrix = np.column_stack([
            np.ones(len(X_sample)),
            X_sample
        ])

        beta = np.linalg.lstsq(
            X_matrix,
            y_sample,
            rcond=None
        )[0]

        beta_results.append(beta[1:])

    beta_results = np.array(beta_results)

    return beta_results


# -------- FMB Model --------

boot_FMB = bootstrap_path(
    structural_df[["FL","DFB"]],
    structural_df["FMB"]
)


print("FMB Bootstrap Complete")
```

```python
import numpy as np
import pandas as pd


def calculate_r2(X, y):
    X_matrix = np.column_stack([np.ones(len(X)), X])
    
    beta = np.linalg.lstsq(X_matrix, y, rcond=None)[0]
    
    y_pred = X_matrix @ beta
    
    ss_res = ((y - y_pred)**2).sum()
    ss_tot = ((y - y.mean())**2).sum()
    
    return 1 - (ss_res / ss_tot)



def calculate_f2(full_r2, excluded_r2):
    return (full_r2 - excluded_r2) / (1 - full_r2)



# ---------- FMB ----------

full_FMB_r2 = calculate_r2(
    structural_df[["FL","DFB"]],
    structural_df["FMB"]
)

f2_FL_FMB = calculate_f2(
    full_FMB_r2,
    calculate_r2(
        structural_df[["DFB"]],
        structural_df["FMB"]
    )
)

f2_DFB_FMB = calculate_f2(
    full_FMB_r2,
    calculate_r2(
        structural_df[["FL"]],
        structural_df["FMB"]
    )
)



# ---------- ISB ----------

full_ISB_r2 = calculate_r2(
    structural_df[["DFB","SI"]],
    structural_df["ISB"]
)

f2_DFB_ISB = calculate_f2(
    full_ISB_r2,
    calculate_r2(
        structural_df[["SI"]],
        structural_df["ISB"]
    )
)

f2_SI_ISB = calculate_f2(
    full_ISB_r2,
    calculate_r2(
        structural_df[["DFB"]],
        structural_df["ISB"]
    )
)



# ---------- FFO ----------

full_FFO_r2 = calculate_r2(
    structural_df[["FL","FMB","ISB"]],
    structural_df["FFO"]
)


f2_FL_FFO = calculate_f2(
    full_FFO_r2,
    calculate_r2(
        structural_df[["FMB","ISB"]],
        structural_df["FFO"]
    )
)


f2_FMB_FFO = calculate_f2(
    full_FFO_r2,
    calculate_r2(
        structural_df[["FL","ISB"]],
        structural_df["FFO"]
    )
)


f2_ISB_FFO = calculate_f2(
    full_FFO_r2,
    calculate_r2(
        structural_df[["FL","FMB"]],
        structural_df["FFO"]
    )
)



f2_results = pd.DataFrame({
    "Path":[
        "FL → FMB",
        "DFB → FMB",
        "DFB → ISB",
        "SI → ISB",
        "FL → FFO",
        "FMB → FFO",
        "ISB → FFO"
    ],
    "f²":[
        f2_FL_FMB,
        f2_DFB_FMB,
        f2_DFB_ISB,
        f2_SI_ISB,
        f2_FL_FFO,
        f2_FMB_FFO,
        f2_ISB_FFO
    ]
})


f2_results
```

```python
import numpy as np
import pandas as pd
import math


def mediation_bootstrap(X, M, Y, n_boot=5000):

    indirect_effects = []

    n = len(X)

    for i in range(n_boot):

        idx = np.random.randint(0, n, n)

        X_sample = X.iloc[idx].values.flatten()
        M_sample = M.iloc[idx].values
        Y_sample = Y.iloc[idx].values


        # Path a: X → M
        X_matrix = np.column_stack([
            np.ones(len(X_sample)),
            X_sample
        ])

        beta_a = np.linalg.lstsq(
            X_matrix,
            M_sample,
            rcond=None
        )[0][1]


        # Path b: M → Y (controlling X)

        MX_matrix = np.column_stack([
            np.ones(len(X_sample)),
            M_sample,
            X_sample
        ])

        beta_b = np.linalg.lstsq(
            MX_matrix,
            Y_sample,
            rcond=None
        )[0][1]


        indirect_effects.append(beta_a * beta_b)


    return np.array(indirect_effects)



# M1: DFB → FMB → FFO

indirect_M1 = mediation_bootstrap(
    structural_df["DFB"],
    structural_df["FMB"],
    structural_df["FFO"]
)


print("M1 Complete")
```

```python
import numpy as np


def calculate_q2(y_true, y_pred):
    ss_res = ((y_true - y_pred)**2).sum()
    ss_tot = ((y_true - y_true.mean())**2).sum()
    
    return 1 - (ss_res / ss_tot)


# FMB Q²

X_FMB = np.column_stack([
    np.ones(len(structural_df)),
    structural_df["FL"],
    structural_df["DFB"]
])

beta_FMB = np.linalg.lstsq(
    X_FMB,
    structural_df["FMB"],
    rcond=None
)[0]

pred_FMB = X_FMB @ beta_FMB

Q2_FMB = calculate_q2(
    structural_df["FMB"],
    pred_FMB
)


# ISB Q²

X_ISB = np.column_stack([
    np.ones(len(structural_df)),
    structural_df["DFB"],
    structural_df["SI"]
])

beta_ISB = np.linalg.lstsq(
    X_ISB,
    structural_df["ISB"],
    rcond=None
)[0]

pred_ISB = X_ISB @ beta_ISB

Q2_ISB = calculate_q2(
    structural_df["ISB"],
    pred_ISB
)


# FFO Q²

X_FFO = np.column_stack([
    np.ones(len(structural_df)),
    structural_df["FL"],
    structural_df["FMB"],
    structural_df["ISB"]
])

beta_FFO = np.linalg.lstsq(
    X_FFO,
    structural_df["FFO"],
    rcond=None
)[0]

pred_FFO = X_FFO @ beta_FFO

Q2_FFO = calculate_q2(
    structural_df["FFO"],
    pred_FFO
)


print("Q² FMB =", round(Q2_FMB,3))
print("Q² ISB =", round(Q2_ISB,3))
print("Q² FFO =", round(Q2_FFO,3))
```

```python


# ============================================================
# SUPPLEMENTARY TABLE S1
# Complete Pairwise Comparisons Across Academic Disciplines
# Compatible with VS Code, Jupyter Notebook and local Python
#
# INPUT:
#   04_pairwise_mann_whitney.csv
#
# OUTPUT:
#   Supplementary_Table_S1.docx
#   Supplementary_Table_S1.xlsx
#   Supplementary_Table_S1.csv
#
# IMPORTANT:
# This code formats source results; it does not recalculate
# the statistical tests or adjusted p-values.
# ============================================================

from pathlib import Path
import re
import sys
import subprocess
import importlib.util

# ------------------------------------------------------------
# 1. INSTALL REQUIRED PACKAGES
# ------------------------------------------------------------

packages = {
    "pandas": "pandas",
    "python-docx": "docx",
    "openpyxl": "openpyxl",
}

for package, module in packages.items():
    if importlib.util.find_spec(module) is None:
        print(f"Installing {package}...")
        subprocess.check_call(
            [sys.executable, "-m", "pip", "install", package]
        )

import pandas as pd

from docx import Document
from docx.shared import Inches, Pt, RGBColor
from docx.enum.section import WD_ORIENT
from docx.enum.text import WD_ALIGN_PARAGRAPH
from docx.enum.table import WD_TABLE_ALIGNMENT, WD_CELL_VERTICAL_ALIGNMENT
from docx.oxml import OxmlElement
from docx.oxml.ns import qn

from openpyxl.styles import Font, Alignment, PatternFill
from openpyxl.utils import get_column_letter


# ------------------------------------------------------------
# 2. FIND THE INPUT CSV
# ------------------------------------------------------------

# If running a .py script, this uses the script's folder.
# If running a Jupyter notebook, it uses the current folder.
try:
    BASE_DIR = Path(__file__).resolve().parent
except NameError:
    BASE_DIR = Path.cwd()

INPUT_FILE = BASE_DIR / "04_pairwise_mann_whitney.csv"

# If your CSV is elsewhere, replace the line above with:
# INPUT_FILE = Path(r"C:\YourFolder\04_pairwise_mann_whitney.csv")

if not INPUT_FILE.exists():
    raise FileNotFoundError(
        "\nCSV file not found.\n"
        f"Python is looking here:\n{INPUT_FILE}\n\n"
        "Place 04_pairwise_mann_whitney.csv in this folder:\n"
        f"{BASE_DIR}\n"
        "Or edit INPUT_FILE to provide the exact CSV path."
    )

print("Reading source file:")
print(INPUT_FILE)


# ------------------------------------------------------------
# 3. LOAD THE SOURCE DATA
# ------------------------------------------------------------

df = pd.read_csv(INPUT_FILE, encoding="utf-8-sig")
df.columns = [str(c).strip() for c in df.columns]

if df.empty:
    raise ValueError("The CSV file is empty.")

print("\nSource rows:", len(df))
print("Source columns:")
print(df.columns.tolist())


# ------------------------------------------------------------
# 4. MATCH SOURCE COLUMN NAMES
# ------------------------------------------------------------

def normalize(name):
    return re.sub(r"[^a-z0-9]", "", str(name).lower())


available = {normalize(c): c for c in df.columns}

aliases = {
    "Outcome": [
        "outcome", "construct", "variable", "measure",
        "dependent_variable", "dependent variable"
    ],
    "Group A": [
        "group_a", "groupa", "group_1", "group1",
        "group_a_label", "discipline_a"
    ],
    "Group B": [
        "group_b", "groupb", "group_2", "group2",
        "group_b_label", "discipline_b"
    ],
    "n_A": [
        "n_a", "na", "n_group_a", "n1", "group_a_n"
    ],
    "n_B": [
        "n_b", "nb", "n_group_b", "n2", "group_b_n"
    ],
    "Mean A": [
        "mean_a", "meana", "group_a_mean",
        "mean_group_a", "mean_1"
    ],
    "Mean B": [
        "mean_b", "meanb", "group_b_mean",
        "mean_group_b", "mean_2"
    ],
    "U": [
        "u", "u_statistic", "mann_whitney_u",
        "mann_whitney_u_statistic"
    ],
    "Rank-biserial r": [
        "rank_biserial_r", "rank_biserial",
        "rank_biserial_correlation", "r_rb", "rrb"
    ],
    "Raw p": [
        "raw_p", "p_raw", "p_value_raw",
        "pvalue_raw", "raw_p_value", "p"
    ],
    "Holm-adjusted p": [
        "holm_p", "p_holm", "p_adj_holm",
        "holm_adjusted_p", "p_holm_adjusted"
    ],
    "BH-adjusted p": [
        "bh_p", "p_bh", "p_adj_bh",
        "bh_adjusted_p", "p_bh_adjusted",
        "fdr_p", "p_fdr"
    ],
}

column_map = {}
missing_columns = []

for target, candidates in aliases.items():
    source_col = None

    for candidate in candidates:
        key = normalize(candidate)
        if key in available:
            source_col = available[key]
            break

    if source_col is None:
        missing_columns.append(target)
    else:
        column_map[target] = source_col

if missing_columns:
    raise ValueError(
        "\nCould not match these required columns:\n"
        f"{missing_columns}\n\n"
        "Actual CSV columns are:\n"
        f"{df.columns.tolist()}\n\n"
        "Edit the aliases dictionary in Section 4 to match "
        "your actual CSV column names."
    )

table = df[
    [column_map[col] for col in aliases.keys()]
].copy()

table.columns = list(aliases.keys())


# ------------------------------------------------------------
# 5. VALIDATE THE DATA
# ------------------------------------------------------------

if table.isna().any().any():
    missing = table.columns[table.isna().any()].tolist()
    raise ValueError(
        f"Missing values found in required columns: {missing}. "
        "Check the source CSV before producing the final table."
    )

numeric_columns = [
    "n_A", "n_B", "Mean A", "Mean B", "U",
    "Rank-biserial r", "Raw p",
    "Holm-adjusted p", "BH-adjusted p"
]

for col in numeric_columns:
    table[col] = pd.to_numeric(table[col], errors="raise")

for col in ["n_A", "n_B"]:
    if (table[col] <= 0).any():
        raise ValueError(f"{col} must contain positive sample sizes.")

for col in ["Raw p", "Holm-adjusted p", "BH-adjusted p"]:
    if not table[col].between(0, 1).all():
        raise ValueError(f"{col} contains values outside [0, 1].")

if len(table) != 36:
    print(
        f"\nWARNING: Expected 36 comparisons; found {len(table)}."
        "\nVerify that the source file contains all comparisons."
    )

counts = table.groupby("Outcome", sort=False).size()

print("\nComparisons by outcome:")
print(counts.to_string())

if table["Outcome"].nunique() != 6:
    print(
        "\nWARNING: The source does not contain exactly "
        "six distinct outcomes. Verify the labels."
    )

# Preserve source order and source values.
# No missing statistics are invented or imputed.


# ------------------------------------------------------------
# 6. FORMAT DISPLAY VALUES
# ------------------------------------------------------------

def format_number(value, digits=3):
    return f"{float(value):.{digits}f}"


def format_p(value):
    value = float(value)
    if value < 0.001:
        return "<.001"
    return f"{value:.3f}".lstrip("0")


display = table.copy()

for col in [
    "Mean A", "Mean B", "U", "Rank-biserial r"
]:
    display[col] = display[col].map(
        lambda value: format_number(value, 3)
    )

for col in ["Raw p", "Holm-adjusted p", "BH-adjusted p"]:
    display[col] = display[col].map(format_p)

display["n_A"] = table["n_A"].astype(int).astype(str)
display["n_B"] = table["n_B"].astype(int).astype(str)


# ------------------------------------------------------------
# 7. DEFINE OUTPUT FILES
# ------------------------------------------------------------

OUTPUT_DIR = INPUT_FILE.parent

DOCX_FILE = OUTPUT_DIR / "Supplementary_Table_S1.docx"
XLSX_FILE = OUTPUT_DIR / "Supplementary_Table_S1.xlsx"
CSV_FILE = OUTPUT_DIR / "Supplementary_Table_S1.csv"


# ------------------------------------------------------------
# 8. EXPORT COMPLETE CSV
# ------------------------------------------------------------

table.to_csv(CSV_FILE, index=False, encoding="utf-8-sig")


# ------------------------------------------------------------
# 9. EXPORT EXCEL WORKBOOK
# ------------------------------------------------------------

with pd.ExcelWriter(XLSX_FILE, engine="openpyxl") as writer:

    table.to_excel(
        writer,
        sheet_name="All comparisons",
        index=False
    )

    ws = writer.sheets["All comparisons"]
    ws.freeze_panes = "A2"
    ws.auto_filter.ref = ws.dimensions

    for cell in ws[1]:
        cell.font = Font(bold=True, color="FFFFFF")
        cell.fill = PatternFill("solid", fgColor="24476B")
        cell.alignment = Alignment(
            horizontal="center",
            vertical="center",
            wrap_text=True
        )

    for col_idx, column in enumerate(ws.columns, start=1):
        max_length = max(
            len(str(cell.value or ""))
            for cell in column
        )
        ws.column_dimensions[
            get_column_letter(col_idx)
        ].width = min(max(max_length + 2, 12), 32)

    # Optional separate sheets for each outcome
    for outcome, subset in table.groupby("Outcome", sort=False):

        sheet_name = re.sub(
            r"[\[\]:*?/\\]", "", str(outcome)
        )[:31] or "Outcome"

        subset.to_excel(
            writer,
            sheet_name=sheet_name,
            index=False
        )


# ------------------------------------------------------------
# 10. CREATE FORMATTED WORD DOCUMENT
# ------------------------------------------------------------

doc = Document()
section = doc.sections[0]

# Landscape layout helps accommodate the 12-column table
section.orientation = WD_ORIENT.LANDSCAPE
section.page_width, section.page_height = (
    section.page_height,
    section.page_width
)

section.top_margin = Inches(0.55)
section.bottom_margin = Inches(0.55)
section.left_margin = Inches(0.45)
section.right_margin = Inches(0.45)

doc.styles["Normal"].font.name = "Times New Roman"
doc.styles["Normal"].font.size = Pt(9)

heading = doc.add_paragraph()
heading.alignment = WD_ALIGN_PARAGRAPH.CENTER

run = heading.add_run(
    "Supplementary Table S1. Complete Pairwise Comparisons "
    "Across Academic Disciplines"
)
run.bold = True
run.font.name = "Times New Roman"
run.font.size = Pt(12)


headers = [
    "Outcome",
    "Group A",
    "Group B",
    "n_A",
    "n_B",
    "Mean A",
    "Mean B",
    "U",
    "Rank-biserial r",
    "Raw p",
    "Holm-adjusted p",
    "BH-adjusted p",
]

word_table = doc.add_table(rows=1, cols=len(headers))
word_table.style = "Table Grid"
word_table.alignment = WD_TABLE_ALIGNMENT.CENTER
word_table.autofit = True


def shade_cell(cell, fill_color):
    tc_pr = cell._tc.get_or_add_tcPr()
    shading = OxmlElement("w:shd")
    shading.set(qn("w:fill"), fill_color)
    tc_pr.append(shading)


def set_repeat_table_header(row):
    tr_pr = row._tr.get_or_add_trPr()
    header_tag = OxmlElement("w:tblHeader")
    header_tag.set(qn("w:val"), "true")
    tr_pr.append(header_tag)


header_row = word_table.rows[0]
set_repeat_table_header(header_row)

for idx, header in enumerate(headers):
    cell = header_row.cells[idx]
    cell.text = header
    shade_cell(cell, "24476B")
    cell.vertical_alignment = WD_CELL_VERTICAL_ALIGNMENT.CENTER

    for paragraph in cell.paragraphs:
        paragraph.alignment = WD_ALIGN_PARAGRAPH.CENTER

        for run in paragraph.runs:
            run.bold = True
            run.font.name = "Times New Roman"
            run.font.size = Pt(7)
            run.font.color.rgb = RGBColor(255, 255, 255)


# Add all comparisons
for _, row in display.iterrows():

    cells = word_table.add_row().cells

    values = [
        row["Outcome"],
        row["Group A"],
        row["Group B"],
        row["n_A"],
        row["n_B"],
        row["Mean A"],
        row["Mean B"],
        row["U"],
        row["Rank-biserial r"],
        row["Raw p"],
        row["Holm-adjusted p"],
        row["BH-adjusted p"],
    ]

    for idx, value in enumerate(values):

        cells[idx].text = str(value)
        cells[idx].vertical_alignment = (
            WD_CELL_VERTICAL_ALIGNMENT.CENTER
        )

        for paragraph in cells[idx].paragraphs:

            if idx < 3:
                paragraph.alignment = WD_ALIGN_PARAGRAPH.LEFT
            else:
                paragraph.alignment = WD_ALIGN_PARAGRAPH.CENTER

            for run in paragraph.runs:
                run.font.name = "Times New Roman"
                run.font.size = Pt(7)


# Table note
note = doc.add_paragraph()
note.paragraph_format.space_before = Pt(5)

note_text = (
    "Note. n_A and n_B denote the sample sizes of Groups A and B, "
    "respectively; U = Mann–Whitney U statistic; r_rb = rank-biserial "
    "correlation; BH = Benjamini–Hochberg. Adjusted p-values are "
    "reproduced from the source CSV. Verify the multiple-testing "
    "correction families against the original analysis code. "
    "Displayed p-values below .001 are reported as <.001."
)

run = note.add_run(note_text)
run.font.name = "Times New Roman"
run.font.size = Pt(8)
run.italic = True

doc.save(DOCX_FILE)


# ------------------------------------------------------------
# 11. FINAL REPORT
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("SUPPLEMENTARY TABLE S1 CREATED SUCCESSFULLY")
print("=" * 60)

print("\nInput:")
print(INPUT_FILE)

print("\nOutput files:")

for path in [DOCX_FILE, XLSX_FILE, CSV_FILE]:
    print(f"\n{path.name}")
    print(f"Location: {path}")
    print(f"Exists: {path.exists()}")

print("\nRows in final table:", len(table))
print("Distinct outcomes:", table["Outcome"].nunique())

print(
    "\nNext step: Open the Word document and verify that "
    "the table fits the page and all labels are readable."
)
```

```python

from pathlib import Path
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch

# ==========================================
# PUBLICATION FIGURE 1: CONCEPTUAL FRAMEWORK
# ==========================================

OUT = Path(r"C:\Users\user\OneDrive\Desktop\pls research")
OUT.mkdir(parents=True, exist_ok=True)

NAME = "Figure_1_Conceptual_Framework"

plt.rcParams.update({
    "font.family": "DejaVu Sans",
    "svg.fonttype": "none",
    "pdf.fonttype": 42,
})

fig, ax = plt.subplots(figsize=(16, 9), dpi=160)
fig.patch.set_facecolor("white")
ax.set_facecolor("white")
ax.set_xlim(0, 16)
ax.set_ylim(0, 9)
ax.axis("off")

# Publication-friendly color palette
TEXT = "#172B3A"
NAVY = "#24476B"

BLUE_FILL, BLUE_EDGE = "#E5F0FF", "#3977B8"
TEAL_FILL, TEAL_EDGE = "#DDF4EC", "#23866D"
PEACH_FILL, PEACH_EDGE = "#FFF0DF", "#C77925"
PURPLE_FILL, PURPLE_EDGE = "#F0E8FF", "#7957B5"

# x, y, width, height, fill, edge, title, subtitle
nodes = {
    "FL": (
        0.55, 5.90, 3.45, 1.30,
        BLUE_FILL, BLUE_EDGE,
        "Financial Literacy", "(FL)"
    ),
    "DFB": (
        0.55, 3.45, 3.45, 1.30,
        BLUE_FILL, BLUE_EDGE,
        "Digital Financial", "Behavior (DFB)"
    ),
    "SI": (
        0.55, 1.00, 3.45, 1.30,
        BLUE_FILL, BLUE_EDGE,
        "Social Influence", "(SI)"
    ),
    "FMB": (
        5.35, 5.80, 3.75, 1.40,
        TEAL_FILL, TEAL_EDGE,
        "Financial Management", "Behavior (FMB)"
    ),
    "ISB": (
        5.35, 3.15, 3.75, 1.40,
        TEAL_FILL, TEAL_EDGE,
        "Impulsive Spending", "Behavior (ISB)"
    ),
    "FFO": (
        11.10, 4.35, 4.15, 1.50,
        PEACH_FILL, PEACH_EDGE,
        "Future Financial", "Orientation (FFO)"
    ),
}

# Draw construct boxes
for key, (x, y, w, h, fill, edge, title, subtitle) in nodes.items():
    ax.add_patch(FancyBboxPatch(
        (x, y), w, h,
        boxstyle="round,pad=0.045,rounding_size=0.18",
        facecolor=fill,
        edgecolor=edge,
        linewidth=2.5,
        zorder=3
    ))

    ax.text(
        x + w/2, y + h*0.60, title,
        ha="center", va="center",
        fontsize=16, fontweight="bold",
        color=TEXT, zorder=4
    )

    ax.text(
        x + w/2, y + h*0.34, subtitle,
        ha="center", va="center",
        fontsize=16, fontweight="bold",
        color=TEXT, zorder=4
    )

# Arrow function
def draw_path(start, end, hypothesis, rad=0, label_xy=None):
    ax.add_patch(FancyArrowPatch(
        start, end,
        arrowstyle="-|>",
        mutation_scale=22,
        linewidth=2.5,
        color=NAVY,
        connectionstyle=f"arc3,rad={rad}",
        shrinkA=4,
        shrinkB=6,
        zorder=2
    ))

    if label_xy is not None:
        ax.text(
            label_xy[0], label_xy[1], hypothesis,
            ha="center", va="center",
            fontsize=15, fontweight="bold",
            color=NAVY,
            bbox=dict(
                boxstyle="round,pad=0.16",
                facecolor="white",
                edgecolor="none",
                alpha=0.98
            ),
            zorder=5
        )

# H1: DFB -> FMB
draw_path(
    (4.00, 4.50), (5.35, 6.05),
    "H1", label_xy=(4.65, 5.30)
)

# H2: DFB -> ISB
draw_path(
    (4.00, 3.95), (5.35, 3.75),
    "H2", label_xy=(4.65, 4.05)
)

# H3: FL -> FMB
draw_path(
    (4.00, 6.55), (5.35, 6.45),
    "H3", label_xy=(4.65, 6.80)
)

# H4: FL -> FFO
# Curved path routed above FMB, not across its text
draw_path(
    (3.10, 7.20), (12.65, 5.85),
    "H4", rad=-0.28, label_xy=(7.75, 8.12)
)

# H5: FMB -> FFO
draw_path(
    (9.10, 6.30), (11.10, 5.25),
    "H5", label_xy=(10.10, 6.05)
)

# H6: ISB -> FFO
draw_path(
    (9.10, 3.85), (11.10, 4.65),
    "H6", label_xy=(10.10, 4.05)
)

# H7: SI -> ISB
draw_path(
    (4.00, 1.60), (5.35, 3.40),
    "H7", label_xy=(4.60, 2.40)
)

# Title
ax.text(
    8, 8.72, "Conceptual Framework",
    ha="center", va="center",
    fontsize=29, fontweight="bold",
    color=TEXT
)

# Figure note
ax.text(
    8, 0.40,
    "Note: H1–H7 represent the seven hypothesized direct relationships.",
    ha="center", va="center",
    fontsize=13, color=NAVY
)

fig.tight_layout(pad=1.5)

# Export in three publication formats
for ext in ("png", "pdf", "svg"):
    path = OUT / f"{NAME}.{ext}"
    options = {
        "bbox_inches": "tight",
        "facecolor": "white",
        "pad_inches": 0.15
    }

    if ext == "png":
        options["dpi"] = 600

    fig.savefig(path, **options)
    print(f"Saved: {path}")

plt.show()
plt.close(fig)

print("\nAll figure files exported successfully.")
```
